Data types in python

For storing number python has three main data types
1. integer - int (for storing real whole numbers, eg= 10, 11, -12)
2. floating point - (for strong real floating point numbers eg 10.1, -11.2 )
3. complex - (for storing complex number eg= 3 + 3j )

```python
number= 1
print(type(number))
number= 1.1
print(type(number))
number= 2 + 2j
print(type(number))
```

List
--------
list()
mutable data type/ mutable data structure

so if we want to store mutilple values using single name we need to use list in python.

so the name of the list will give us the first location of the array

to access the rest of the values we've to use subscripts where we give index of each cell of the array
eg. list[0] - for accessing first element
    list[1] - for accessing second element

so when we give the subscript with 0th index the first location will be accessed so for the next location

first location + ( sizeof(data type) * index)

100 + (4 * 2)= 108 (this is for integer and the index is 2, list[2])


```python
list_1=[14, 1, 1.2, "some name"]

'''
her we are discussing about lists in python
every index of python list will be starting from 0 to size/length - 1

'''
print(f"this is the 0th index value {list_1[0]}")
print(f"this is the 1th index value {list_1[1]}")
print(f"this is the 2th index value {list_1[2]}")
print(f"this is the 3th index value {list_1[3]}")
#we are using len function to get the size or length of the array
print(f"length/size of the array is {len(list_1)}")
```

```python
#printing the entire list
print(list_1)
```

```python
#updating the second index of the list with the value 10
list_1[2]= 10
```

```python
#again printing the updated list
print(list_1)
```
