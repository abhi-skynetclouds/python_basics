identity operators is python 
is operator a is b
is not operator a is not b

```python
x = [1, 2, 3]
y = [1, 2, 4]
z = x

print(x == y)  # True (equal content)
print(x is y)   # False (different objects in memory)
print(x is z)   # True (same object reference)
print(x is not y) # False (different object reference)

print(x)
print(z)
print(y)
```
